```python
# Capstone evaluation worksheet
model_candidates = [
    {"name": "model_a", "quality": 0.0, "latency_ms": 0, "cost_per_1k_tokens": 0.0},
    {"name": "model_b", "quality": 0.0, "latency_ms": 0, "cost_per_1k_tokens": 0.0}
]

def score_model(m):
    # Example weighted score: higher quality and lower latency/cost are better
    return (0.6 * m["quality"]) - (0.2 * (m["latency_ms"] / 1000)) - (0.2 * m["cost_per_1k_tokens"])

ranked = sorted(model_candidates, key=score_model, reverse=True)
ranked
```
